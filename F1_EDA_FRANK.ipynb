## Formar dataframe con datos que interesan

```python
races_sub = races[['raceId', 'year', 'round', 'circuitId', 'name', 'date']]
races_sub = races_sub.rename(columns={'name': 'race_name'})

df = results.merge(races_sub, on='raceId', how='left')
```

Selecciona columnas útiles de races.  
Renombra name a race_name para evitar choques.  
Une results con races por raceId.

```python
drivers_sub = drivers[['driverId', 'forename', 'surname', 'nationality']]
drivers_sub = drivers_sub.rename(columns={'nationality': 'driver_nationality'})

df = df.merge(drivers_sub, on='driverId', how='left')
```

Selecciona nombre y nacionalidad del piloto.  
Renombra nationality a driver_nationality.  
Une el resultado con drivers por driverId.

```python
constructors_sub = constructors[['constructorId', 'name', 'nationality']]
constructors_sub = constructors_sub.rename(columns={
    'name': 'constructor_name',
    'nationality': 'constructor_nationality'
})

df = df.merge(constructors_sub, on='constructorId', how='left')
```

Selecciona nombre y nacionalidad del equipo.  
Renombra ambas columnas para distinguirlas del piloto.  
Une el resultado con constructors por constructorId.

```python
df = df.merge(status, on='statusId', how='left')
```

Une el resultado con status por statusId.  
Traduce el código numérico a texto legible.

```python
pit_stops['duration'] = pd.to_numeric(pit_stops['duration'], errors='coerce')

pit_summary = pit_stops.groupby(['raceId', 'driverId']).agg(
    num_stops=('stop', 'count'),
    total_pit_time=('duration', 'sum')
).reset_index()

df = df.merge(pit_summary, on=['raceId', 'driverId'], how='left')
```

Convierte duration a tipo numérico.  
Agrupa pit_stops por carrera y piloto.  
Calcula cantidad de paradas y tiempo total en boxes.  
Une el resumen al DataFrame principal.

```python
df.head()
```

Muestra las primeras filas del DataFrame unido.  
Confirma que las uniones funcionaron correctamente.  
num_stops y total_pit_time quedan nulos antes de 2011.

## Limpieza de Datos

Calcula el porcentaje de nulos por columna.  
Ordena las columnas de mayor a menor.

```python
nulos = df.isnull().mean().sort_values(ascending=False) * 100
nulos[nulos > 0]
```

results.csv marca valores faltantes con "\N".  
Convierte esas columnas numéricas a tipo numérico.  
Los valores no convertibles quedan como NaN.

```python
cols_numericas = ['position', 'milliseconds', 'fastestLap', 'rank', 'fastestLapSpeed']

for col in cols_numericas:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

pit_stops no registra carreras previas a 2011.  
Marca esas filas antes de rellenar con una columna aparte.  
Rellena num_stops y total_pit_time con cero.

```python
df['tiene_datos_pit'] = df['num_stops'].notna()

df['num_stops'] = df['num_stops'].fillna(0)
df['total_pit_time'] = df['total_pit_time'].fillna(0)
```

Verifica duplicados usando resultId.  
resultId es la clave única de la tabla results.
